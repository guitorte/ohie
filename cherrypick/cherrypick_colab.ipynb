# 🍒 Cherrypick on Google Colab

Bulk content extractor that lets you pick the pages first: **discover** the links on a site
(with sub-pages grouped under their parent), **tick** the ones you want, then **extract**
them to Markdown as a `.zip`.

Run the two cells below in order (**Runtime → Run all** works too). The app opens right
under the second cell, and a public `gradio.live` link is printed as well, in case you
prefer a full browser tab.

## 1 · Install and fetch the app

```python
# Branch of github.com/guitorte/ohie to load Cherrypick from.
BRANCH = "ccr-614e7029-ryf7bw"   # switch to "main" once the branch is merged

!pip install -q -U "gradio>=6.0" requests beautifulsoup4 markdownify

import os, urllib.request
os.makedirs("/content/cherrypick", exist_ok=True)
for name in ("app.py", "crawler.py"):
    url = f"https://raw.githubusercontent.com/guitorte/ohie/{BRANCH}/cherrypick/{name}"
    urllib.request.urlretrieve(url, f"/content/cherrypick/{name}")
print("Ready:", os.listdir("/content/cherrypick"))
```

## 2 · Launch

Keep this cell running while you use the app; stop it (■) to shut the app down.
